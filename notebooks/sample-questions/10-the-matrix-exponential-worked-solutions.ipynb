# Section 10 · The matrix exponential and the expected order flow

**F1. The matrix $K$ and its eigenvalues.**
$K = A - \beta I$, Proposition 1.3.27 (`prop.hurwitzCriterion`), is $\beta(\Gamma - I)$:
it has the eigenvectors of $\Gamma = A/\beta$, with the eigenvalues $\beta(\gamma_i - 1)$
over the eigenvalues $\gamma_i$ of $\Gamma$, as in the proof of that proposition.
- On the self- and cross-exciting pairs $A$ has equal diagonal entries $a$ and equal off-diagonal entries $b$,
  so $K$ has the eigenvalue $a + b - 4$ on $(1, 1)$ and $a - b - 4$ on $(1, -1)$:
  $-1$ and $-1.4$ on the self-exciting pair, $-1$ and $-6.6$ on the cross-exciting pair.
- On the one-way pair $K = \begin{pmatrix} -3.2 & 0 \\ 4 & -1.6 \end{pmatrix}$ is triangular,
  with the eigenvalue $-1.6$ on $(0, 1)$ and $-3.2$ on $(0.4, -1)$.
  A buy excites the sells, $A_{2,1} = 4$, and no sell excites a buy, $A_{1,2} = 0$:
  the row is excited and the column excites, Remark 1.3.7 (`remark.indexOrder`).

$\Gamma = I + K/\beta$ has the eigenvalues $0.75, 0.65$ on the self-exciting pair,
$0.75, -0.65$ on the cross-exciting pair and $0.6, 0.2$ on the one-way pair:
$\rho = 0.75$, $0.75$ and $0.6$, all below 1.

**F2. Functions of $K$ on its eigenvectors.**
If $Kv = \zeta v$, then $e^{Ku}v = e^{\zeta u}v$ and
$$\Phi(h)\,v = \int_0^h e^{\zeta u}\,du\; v = \frac{e^{\zeta h} - 1}{\zeta}\,v,
\qquad \Phi(h)\,v = h\,v \text{ when } \zeta = 0.$$
With eigenvectors as the columns of an invertible $P$,
$e^{Kh} = P\,\mathrm{diag}(e^{\zeta h})\,P^{-1}$.
On the self- and cross-exciting pairs, with
$\Pi_\pm = \frac12\begin{pmatrix} 1 & \pm 1 \\ \pm 1 & 1 \end{pmatrix}$
the projections on $(1, \pm 1)$ and $\zeta_\pm$ the eigenvalues there,
$$e^{Kh} = e^{\zeta_+ h}\,\Pi_+ + e^{\zeta_- h}\,\Pi_-,$$
whose first column is $\tfrac12\big(e^{\zeta_+ h} + e^{\zeta_- h},\; e^{\zeta_+ h} - e^{\zeta_- h}\big)$.
These are facts of linear algebra.

**F3. Non-negativity and commutation.**
When $A \geq 0$, every term of $e^{Ah} = \sum_{k \geq 0}(Ah)^k/k!$ is a non-negative matrix.
Since $\beta I$ commutes with $A$, $e^{Kh} = e^{-\beta h}e^{Ah} \geq 0$,
and so $\Phi(h) = \int_0^h e^{Ku}\,du \geq 0$, with a positive diagonal for $h > 0$.
$A$ commutes with $K$, hence with every $e^{Ku}$ and with $\Phi(h)$.
These too are facts of linear algebra.

**F4. The forward means.**
Given $Z(s+) = z$, Proposition 1.3.33 (`prop.forwardMean`) gives the mean state at $s + h$,
$$m_{s,s+h}(z) = e^{Kh}z + \Phi(h)\,\mu,
\qquad \Phi(h) = \int_0^h e^{Ku}\,du,$$
eq. (1.3.30, `eq.forwardMeanState`), and the expected counts of $(s, s+h]$,
$$h\,\mu + A\,\Phi(h)\,z + A\Big(\int_0^h \Phi(u)\,du\Big)\mu,$$
eq. (1.3.31, `eq.forwardMeanCounts`):
the immigrants, the descendants of the events the state carries, and the descendants of the immigrants.
Both hold at every $\rho$.
When $\rho < 1$, $\Phi(h) = K^{-1}(e^{Kh} - I)$, and with the stationary mean
$m = \lambda^*/\beta$ of eq. (1.3.19, `eq.meanState`) the two read
$$m_{s,s+h}(z) = m + e^{Kh}(z - m)
\qquad\text{and}\qquad
h\,\lambda^* + A\,\Phi(h)(z - m),$$
eq. (1.3.32, `eq.forwardMeanDeviation`) and eq. (1.3.33, `eq.forwardMeanCountsDeviation`).
As $h \to \infty$, $\Phi(h) \to -K^{-1} = (I - \Gamma)^{-1}/\beta$
(the proof of Corollary 1.3.35 (`corol.meanResponse`)), so that
$A\,\Phi(\infty) = \Gamma(I - \Gamma)^{-1} = (I - \Gamma)^{-1} - I$,
Remark 1.4.20 (`remark.totalResponse`).

**F5. The mean response.**
One more event of type $e'$ in the state at $s$ displaces the mean state at $s + h$ by
$$R_{e'}(h) = e^{Kh}\,\mathbf e_{e'},$$
the column $e'$ of $e^{Kh}$, whatever $s$ and $z$,
eq. (1.3.34, `eq.meanResponse`) of Corollary 1.3.35 (`corol.meanResponse`):
the term $\Phi(h)\mu$ is common to the two mean states.
When $\rho < 1$, integrating the displaced total intensity counts the strict descendants,
$$\int_0^\infty \mathbf 1^\top A\,R_{e'}(h)\,dh
= \mathbf 1^\top\big[(I - \Gamma)^{-1} - I\big]\mathbf e_{e'} = C_{e'} - 1,$$
eq. (1.3.35, `eq.responseIntegral`),
with the cluster sizes $C = \mathbf 1^\top(I - \Gamma)^{-1}$ of eq. (1.3.22, `eq.clusterSize`),
which count the ancestor.

**F6. Direction symmetry.**
A specification is direction-symmetric when $\Sigma A \Sigma = A$ and $\Sigma\mu = \mu$,
$\Sigma$ exchanging the buys and the sells, Definition 1.4.12 (`def.directionSymmetry`).
The self- and cross-exciting pairs are; the one-way pair is not.
Then $\varpi^\top M v = 0$ for every $M$ with $\Sigma M \Sigma = M$ and every $v$ with $\Sigma v = v$,
and in particular $\varpi^\top\mu = 0$ and, at $\rho < 1$, $\varpi^\top\lambda^* = 0$,
Proposition 1.4.13 (`prop.directionSymmetry`).
By the paragraph after that proposition, at $\rho < 1$,
$e^{Kh}$, $\Phi(h)$ and $(I - \Gamma)^{-1}$ are such matrices,
and every baseline term contracted with $\varpi$ vanishes.

**F7. Indexing and products in numpy.**
`M[0]` is the first row of `M`, and `M[:, 0]` its first column.
Between a one-dimensional array `v` and a matrix, `v @ M` is $v^\top M$ and `M @ v` is $Mv$:
`ones @ M` gives the column sums of `M`, and `M @ ones` its row sums.
For a stack `R` of shape $(n, 2, 2)$, `R[k, i, j]` is the entry $(i, j)$ of the $k$-th matrix,
and `R.sum(axis=1)` sums over the row index $i$, giving the column sums of each matrix.
These are facts of numpy.

## 10.2 · `question.seriesOnKAgainstSeriesOnTheExcitation`

The cross-exciting pair: the series of $e^{Kh}$ summed to 400 terms,
and $e^{-4h}$ times the series of $e^{Ah}$,
each compared with `expm` at $h = 1$ and at $h = 10$.

**Answer: (B).** `True True`, then `False True`.

Both series converge, and 400 terms are more than either needs at either horizon:
what decides is the rounding.

**The series of $Kh$.**
$K$ has the eigenvalue $-6.6$ on $(1, -1)$ (F1),
so along that eigenvector the terms are $(-6.6h)^k/k!$, alternating in sign.
- At $h = 1$ the largest has the size $6.6^6/6! \approx 115$,
  rounded to within about $10^{-14}$, against entries of about $0.18$: `True`.
- At $h = 10$ the largest has the size $66^k/k! \approx 2\times10^{27}$,
  rounded to within about $2\times10^{27}\cdot10^{-16} = 2\times10^{11}$,
  and the terms cancel to entries of about $2.3\times10^{-5}$.
  The rounding is what is left, far above the tolerance of `np.allclose`: `False`.

**The series of $Ah$.**
$A \geq 0$, so every term $(Ah)^k/k!$ is a non-negative matrix (F3),
and every entry of the sum is a sum of non-negative numbers.
Such a sum loses no digit to cancellation, whatever its size:
its relative error is of the order of $10^{-16}$.
At $h = 10$ it is of the order of $e^{30} \approx 10^{13}$,
from the eigenvalue $3$ of $A$,
and the product with $e^{-40}$ costs one more rounding.
What the sum cannot resolve is the mode $e^{-66}$ on $(1, -1)$, below $10^{-28}$.
`True` at both horizons.

```python
6.6**6 / factorial(6), 2e27 * 1e-16, 0.5 * (exp(-10) + exp(-66)), 0.5 * exp(30) * exp(-40), 0.5 * exp(30)
```

**The other choices.**
- (A) `True True` twice: takes a convergent series to give its sum in floating point;
  at $h = 10$ the terms of the series of $Kh$ cancel over 32 orders of magnitude.
- (C) `True False` on the second line: judges the loss of digits by the size of the sum,
  about $2.3\times10^{-5}$ for the series of $Kh$ and $5\times10^{12}$ for that of $Ah$.
  Cancellation among the terms decides: the series of $Ah$ has none, and that of $Kh$ cancels terms of $10^{27}$.
- (D) `False False` on the second line: takes the product of a sum of the order of $e^{30}$ by $e^{-40}$ to lose digits;
  a product costs one rounding.
- (E) `False True` twice: takes any cancellation for a failure.
  At $h = 1$ the terms reach about 115 along $(1, -1)$, against entries of about $0.18$:
  the cancellation costs some three of the sixteen digits.

## 10.3 · `question.eigenvectorsTransposedForTheInverse`

$e^{0.5K} = P\,\mathrm{diag}(e^{0.5\kappa})\,P^{-1}$,
computed with `np.linalg.inv(P)` and with `P.T` for $P^{-1}$,
on the self-exciting and on the one-way pair.
`np.linalg.eig` returns eigenvectors of unit length,
and the stem writes $\kappa$ for the eigenvalues it returns, after the code's `kappa`.

**Answer: (A).** `True True`, then `True False`.

**By the inverse.**
For any invertible $P$ whose columns are eigenvectors, $P\,\mathrm{diag}(e^{\kappa h})\,P^{-1} = e^{Kh}$ (F2).
Both pairs have two distinct eigenvalues (F1), so $P$ is invertible: `True` on both lines.

**By the transpose.**
$P$ and $\mathrm{diag}(e^{\kappa h})$ are invertible,
so the two products agree only if $P^\top = P^{-1}$,
that is, only if the columns of $P$ are of unit length and orthogonal.
- On the self-exciting pair $K$ is symmetric with distinct eigenvalues,
  so its eigenvectors $(1, \pm 1)/\sqrt 2$ are orthogonal,
  and at unit length $P$ is orthogonal: `True`.
- On the one-way pair the eigenvectors $(0, 1)$ and $(0.4, -1)$ (F1) are not orthogonal:
  their unit vectors $(0, 1)$ and $(0.371, -0.928)$ have the scalar product $-0.928$: `False`.

```python
norm = sqrt(0.4**2 + 1)
0.4 / norm, -1 / norm, 0 * 0.4 / norm + 1 * (-1 / norm)
```

**The other choices.**
- (B) `True True` on both lines: takes unit length for orthonormality.
- (C) `True False` on both lines: takes `P.T` to fail on the self-exciting pair too,
  missing that a symmetric matrix with distinct eigenvalues has orthogonal eigenvectors.
- (D) `False False` on the second line: takes the decomposition itself to need orthogonal eigenvectors;
  $P\,\mathrm{diag}(e^{\kappa h})\,P^{-1}$ needs only $P$ invertible.

- The second row starts at $(0, 1, 0)$:

= 0.25 + 2 \cdot 0.316 \cdot 1 + 2 \cdot 0.092 \cdot 0.5 = 0.974:$$
the immigrants, the descendants of the buy at $s$, and those of the immigrants.

## 10.8 · `question.responseIntegralCountsStrictDescendants`

The one-way pair, $\Gamma = A/4 = \begin{pmatrix} 0.2 & 0 \\ 1 & 0.6 \end{pmatrix}$,
with $\Phi(50)$ for $\Phi(\infty)$.

**Answer: (D).** `[3.375, 1.5]`, `[4.375, 2.5]`, `[0.25, 4.625]`.

**The inverse.**
$I - \Gamma = \begin{pmatrix} 0.8 & 0 \\ -1 & 0.4 \end{pmatrix}$ has the determinant $0.32$, and
$$(I - \Gamma)^{-1} = \frac{1}{0.32}\begin{pmatrix} 0.4 & 0 \\ 1 & 0.8 \end{pmatrix}

## 10.12 · `question.meanResponseOnAClockByEigenvalues`

$e^{Kh}$ at every horizon of a clock of $n$ horizons, on the one-way pair:
one call of `expm` per horizon (A),
against one eigendecomposition followed by operations on arrays (B);
$n = 10^4$ and $n = 10^5$.

**Answer: (B).**
B is faster by a factor of five or more,
and the time of each grows about tenfold from $n = 10^4$ to $n = 10^5$.

- **A** calls `expm` $n$ times from the interpreter.
  Each call scales the matrix, evaluates a rational approximation and squares the result back,
  and pays the fixed cost of a call into scipy, large against the arithmetic of a $2 \times 2$ matrix.
  Its time is that cost times $n$.
- **B** decomposes $K$ once (F2),
  then computes $2n$ scalar exponentials in one `np.exp` and $n$ products of $2 \times 2$ matrices in one `@`.
  These few calls of numpy loop over their $n$ entries in compiled code:
  the fixed cost is paid a few times, not $n$ times, and the work is still proportional to $n$.

Both are linear in $n$: ten times the horizons, about ten times the time.
B is faster by the ratio of the cost of a call of `expm` to the compiled work per horizon.
Runs of the verification notebook print ratios of some 30 to 55,
and growths of about 10 for A and of 6 to 12 for B.

**The other choices.**
- (A) A faster: `expm` is compiled, but A makes its $n$ calls from the interpreter,
  while B builds its stack inside numpy.
- (C) The time of B hardly changes: each of its few calls of numpy works over $n$ entries,
  so its time is proportional to $n$.
- (D) Within a factor of two: both compute $n$ exponentials of a $2 \times 2$ matrix,
  but A pays the fixed cost of a call $n$ times.
