# Section 8 · Telling specifications apart

**F1. Branching matrix and column sums.**
For the exponential kernel $\Gamma = A/\beta$, eq. (1.3.12, `eq.GammaIsAOverBeta`).
$A_{e,e'}$ is the influence of an event of type $e'$ on the intensity of type $e$:
the row is excited and the column excites, Remark 1.3.7 (`remark.indexOrder`).
By Proposition 1.3.8 (`prop.clusterRepresentation`), at $\rho < 1$, an event of type $e'$
has a Poisson number of direct offspring of type $e$, of mean $\Gamma_{e,e'}$,
so the column sum $(\mathbf 1^\top\Gamma)_{e'}$ is the expected number of its direct offspring.
The column sum of $A$ is the jump in the total intensity, larger by the factor $\beta$,
Remark 1.3.10 (`remark.countsAgainstRates`).

**F2. Branching ratio.**
$\rho$ is the spectral radius of $\Gamma$ (Section 1.3.2 (`sec.hawkesProcesses`), after eq. (1.3.9, `eq.branchingMatrix`)),
and a matrix and its transpose share it, Remark 1.3.7 (`remark.indexOrder`).
Lemma 1.3.17 (`lemma.perronFrobenius`), part iii, bounds $\rho$ between the smallest and the largest row sum of $\Gamma$,
and, applied to $\Gamma^\top$, between the smallest and the largest column sum (the paragraph after the lemma):
constant row or column sums $c$ give $\rho = c$,
and a row or a column sum above one does not by itself make $\rho \geq 1$.
The eigenvalues of a $2\times2$ matrix of trace $t$ and determinant $\delta$ are $\big(t \pm \sqrt{t^2 - 4\delta}\big)/2$;
with equal diagonal entries $a$ and off-diagonal entries $b, c \geq 0$ they are $a \pm \sqrt{bc}$;
those of a triangular matrix are its diagonal entries.

**F3. The stationary intensity.**
For baselines $\mu > 0$ entrywise, Theorem 1.3.19 (`thm.stationarity`)
makes the process stationary if and only if $\rho < 1$, and then
$$\lambda^* = (I - \Gamma)^{-1}\mu,$$
eq. (1.3.18, `eq.stationaryIntensity`),
the unique solution of the fixed point $\lambda^* = \mu + \Gamma\lambda^*$ (the paragraph after the theorem).
For a $2\times2$ matrix,
$$\begin{pmatrix} a & b \\ c & d \end{pmatrix}^{-1} = \frac{1}{ad - bc}\begin{pmatrix} d & -b \\ -c & a \end{pmatrix}.$$
When $\mu$ is a right Perron vector, $\Gamma\mu = \rho\mu$,
the solution is $\lambda^* = \mu/(1-\rho)$,
the second of the sufficient conditions stated after Corollary 1.3.25 (`corol.constantClusterSize`).
The total rate is $\bar\lambda = \mathbf 1^\top\lambda^*$ and the total immigration rate $\bar\mu = \mathbf 1^\top\mu$
(the definitions after Remark 1.3.20 (`remark.strictBaseline`)).
The process is then ergodic, and the time averages along one long path estimate these quantities
(the paragraph that precedes eq. (1.3.37, `eq.supercriticalGrowth`)).

**F4. Clusters.**
At $\rho < 1$, by Proposition 1.3.23 (`prop.clusterSize`), the expected size of the cluster founded by an immigrant of type $e'$,
the founder included, is the $e'$-th entry of
$$C = \mathbf 1^\top(I - \Gamma)^{-1},$$
eq. (1.3.22, `eq.clusterSize`): the column sums of $(I - \Gamma)^{-1}$.
In its proof the expected $k$-th generation of the cluster, by type, is $\Gamma^k\mathbf e_{e'}$:
one multiplication by $\Gamma$ per generation, whatever the founder.
The endogenous fraction is $1 - \bar\mu/\bar\lambda$, eq. (1.3.21, `eq.endogenousFraction`).
By Corollary 1.3.25 (`corol.constantClusterSize`), $C_{e'} = 1/(1-\rho)$ for every $e'$
exactly when every column sum of $\Gamma$ is $\rho$.

**F5. Counts in long windows.**
Let
$$D = (I - \Gamma)^{-1}\,\mathrm{diag}(\lambda^*)\,(I - \Gamma)^{-\top},$$
eq. (1.4.20, `eq.flowVolatility`).
The proof of Proposition 1.4.23 (`prop.forecastAtLongHorizons`), through eq. (1.4.21, `eq.countIncrementIdentity`),
uses of its hypotheses only $\rho < 1$ and the stationary version, with $\mu > 0$ through Theorem 1.3.29 (`thm.lyapunov`).
It is made for one linear combination of the counts and holds for every one, without the direction symmetry assumed there:
over a window of length $w$ the counts by type have covariance $wD$ up to a term of order $\sqrt w$.
With constant column sums, $\mathbf 1^\top(I - \Gamma)^{-1} = \mathbf 1^\top/(1-\rho)$ by Corollary 1.3.25 (`corol.constantClusterSize`),
so $\mathbf 1^\top D\,\mathbf 1 = \bar\lambda/(1-\rho)^2$,
and the variance of the total count over its mean tends to $1/(1-\rho)^2$.

**F6. The five specifications.**
All at $\beta = 4$.
- The weakly exciting pair, $\mu = (0.9, 0.9)$:
  $\Gamma = \begin{pmatrix} 0.075 & 0.025 \\ 0.025 & 0.075 \end{pmatrix}$,
  every row and column summing to $0.1 = \rho$;
  $\mu$ is a right Perron vector, so $\lambda^* = \mu/0.9 = (1, 1)$ and $\bar\lambda = 2$.
- The self-exciting pair, $\mu = (0.25, 0.25)$:
  $\Gamma = \begin{pmatrix} 0.7 & 0.05 \\ 0.05 & 0.7 \end{pmatrix}$,
  sums $0.75 = \rho$, $\lambda^* = (1, 1)$, $\bar\lambda = 2$.
- The cross-exciting pair, the self-exciting one with its rows swapped:
  $\Gamma = \begin{pmatrix} 0.05 & 0.7 \\ 0.7 & 0.05 \end{pmatrix}$,
  with the same sums, $\rho$, $\lambda^*$ and $\bar\lambda$.
- The asymmetric pair, $\mu = (0.2, 0.6)$:
  $\Gamma = \begin{pmatrix} 0.3 & 1.2 \\ 0.1 & 0.2 \end{pmatrix}$,
  column sums $(0.4, 1.4)$, row sums $(1.5, 0.3)$;
  trace $0.5$ and determinant $-0.06$ give the eigenvalues $0.6$ and $-0.1$, so $\rho = 0.6$;
  with $\det(I - \Gamma) = 0.44$, $\lambda^* = (2, 1)$ and $\bar\lambda = 3$.
- The one-way pair, $\mu = (0.4, 0.1)$:
  $\Gamma = \begin{pmatrix} 0.2 & 0 \\ 1 & 0.6 \end{pmatrix}$,
  triangular, with eigenvalues $0.2$ and $0.6$, so $\rho = 0.6$; column sums $(1.2, 0.6)$;
  $\lambda^*_1 = 0.4/0.8 = 0.5$, $\lambda^*_2 = (0.1 + 0.5)/0.4 = 1.5$, $\bar\lambda = 2$.

```python
det = 0.7 * 0.8 - 1.2 * 0.1
asymmetric = ((0.8 * 0.2 + 1.2 * 0.6) / det, (0.1 * 0.2 + 0.7 * 0.6) / det)
one_way = (0.4 / 0.8, (0.1 + 1.0 * 0.4 / 0.8) / 0.4)
det, asymmetric, one_way
```

- **S4**: $\Gamma = \begin{pmatrix} 0.3 & 0.1 \\ 0.1 & 0.3 \end{pmatrix}$ has constant sums, so $\rho = 0.4$.

## 8.2 · `question.fixedPointSolvedAtAndAboveCriticality`

$\mu = (0.2, 0.2)$, $\beta = 4$, and three symmetric excitations
with diagonal $2$ and off-diagonal $1.2$, $2.0$ and $2.4$.

**Answer: (A).** `[1. 1.]`, `singular`, `[-2. -2.]`.

$\Gamma = A/4$ has diagonal $0.5$ and off-diagonal $0.3$, $0.5$ and $0.6$:
its rows sum to $0.8$, $1$ and $1.1$, which are $\rho$ (F2).
$\mu$ is proportional to $\mathbf 1$, a right Perron vector of each, $\Gamma\mu = \rho\mu$.
- $\rho = 0.8$: the solution is $\mu/(1-\rho) = 0.2/0.2 = 1$ (F3): `[1. 1.]`.
- $\rho = 1$: $I - \Gamma = \begin{pmatrix} 0.5 & -0.5 \\ -0.5 & 0.5 \end{pmatrix}$ is singular, exactly, in floating point,
  and `np.linalg.solve` raises `LinAlgError`: `singular`.
- $\rho = 1.1$: $I - \Gamma = \begin{pmatrix} 0.5 & -0.6 \\ -0.6 & 0.5 \end{pmatrix}$ has determinant $0.25 - 0.36 = -0.11$
  and is invertible.
  The solution is still $\mu/(1-\rho) = 0.2/(-0.1) = -2$, since $(I - \Gamma)\mu = (1-\rho)\mu$ at every $\rho$: `[-2. -2.]`.

`np.linalg.solve` solves a linear system and checks nothing of $\rho$.
The fixed point is available whatever $\rho$, Remark 1.3.20 (`remark.strictBaseline`),
and for a left Perron vector $v$ the solution $x$ of $(I - \Gamma)x = \mu$ has $v^\top\mu = (1-\rho)\,v^\top x$:
at $\rho > 1$ and $\mu > 0$, $v^\top x < 0$, and $x$ is the intensity of no process.

```python
rows = (2.0 + 1.2) / 4, (2.0 + 2.0) / 4, (2.0 + 2.4) / 4
rows, 0.5 * 0.5 - 0.5 * 0.5, 0.5 * 0.5 - 0.6 * 0.6, 0.2 / (1 - rows[0]), 0.2 / (1 - rows[2])
```

**The other choices.**

- (C) `[2. 2.]` on the third line: the sign lost, $0.2/(1.1 - 1)$ for $0.2/(1 - 1.1)$.

## 8.3 · `question.lookalikeKernelsOneSupercritical`

$\mu = (0.2, 0.2)$, $\beta = 4$,
$A = \begin{pmatrix} 2.0 & 0.4 \\ 3.6 & 2.0 \end{pmatrix}$ and $\begin{pmatrix} 2.0 & 1.6 \\ 3.6 & 2.0 \end{pmatrix}$,

and $0.5 \pm \sqrt{0.36}$, that is $1.1$ and $-0.1$.

= \begin{pmatrix} 0.75 \\ 1.75 \end{pmatrix},
\qquad \bar\lambda = 2.5.$$
Over 4000 s the path has about $\bar\lambda T = 10^4$ events, and `islice` does not bind.
The verification notebook prints 10042.

**The second.**
$\rho = 1.1$.

## 8.4 · `question.buysWithNeitherImmigrantsNorExciters`

$\mu = (0, 0.5)$, $A = \begin{pmatrix} 8 & 0 \\ 2 & 2 \end{pmatrix}$, $\beta = 4$, passed in a `SimpleNamespace`,
over $10^4$ s; the count of each type divided by $10^4$, to one decimal.

**Answer: (D).** It prints `[0. 1.]`.

$\Gamma = A/4 = \begin{pmatrix} 2 & 0 \\ 0.5 & 0.5 \end{pmatrix}$ is triangular, with eigenvalues $2$ and $0.5$:
$\rho = 2$ (F2), which `HawkesParams` would refuse; the namespace passes no check.

**The buys.**
By eq. (1.3.13, `eq.intensityFromState`), $\lambda_1 = \mu_1 + A_{1,1}Z_1 + A_{1,2}Z_2 = 8Z_1$.
$Z_1$ sums over the past buys, eq. (1.3.11, `eq.decayedCounts`), and the path starts empty:
$Z_1 = 0$ until a buy, and a buy needs $\lambda_1 > 0$.
No buy occurs.

**The sells.**
With $Z_1 = 0$, $\lambda_2 = 0.5 + 2Z_2$:
a one-type exponential-kernel process with $\mu = 0.5$ and $\Gamma = 2/4 = 0.5$,
of rate $0.5/(1 - 0.5) = 1$ (F3).
By F5 its rate over $10^4$ s has the standard error $\sqrt{D/T} = \sqrt{4/10^4} = 0.02$,
with $D = \lambda^*/(1-\rho)^2 = 4$, so the printed rate rounds to `1.`.

As in the example of Remark 1.3.20 (`remark.strictBaseline`),
a type with neither immigrants nor an exciter is never reached, and the process is stationary at $\rho = 2$.
Theorem 1.3.19 (`thm.stationarity`) assumes $\mu > 0$ entrywise, which fails here.

```python
rate = 0.5 / (1 - 2 / 4)
rate, 0.5 + 0.5 * 0.5, sqrt(rate / (1 - 2 / 4) ** 2 / 1e4)
```

**The other choices.**
- (A) The run does not end: $\rho = 2$ taken to make the intensity grow;
  Theorem 1.3.19 (`thm.stationarity`) ties stationarity to $\rho < 1$ only when $\mu > 0$ entrywise.
- (B) `[0.  0.5]`: the sells counted as their immigrants alone, $\mu_2 = 0.5$.
- (C) `[0.  0.8]`: the sells counted to their first generation, $\mu_2 + \Gamma_{2,2}\,\mu_2 = 0.75$, which rounds to $0.8$.

## 8.7 · `question.clusterSizeByFounderAndItsTranspose`

For the weakly exciting pair $\lambda_{\mathfrak g}^+ \approx 2.42$ and the Fano factor about $1.004$.

## 8.12 · `question.stationaryIntensityBySeriesAgainstSolve`

$\lambda^*$ by the series $\sum_k\Gamma^k\mu$ to a tolerance of $10^{-12}$ (A) and by `np.linalg.solve` (B),
at $\mu = (0.5, 0.5)$ and $\Gamma = \rho\begin{pmatrix} 0.7 & 0.3 \\ 0.3 & 0.7 \end{pmatrix}$,
for $\rho = 0.9$, $0.99$ and $0.999$.

**Answer: (A).** B is faster at every $\rho$;
the time of A grows about tenfold from one $\rho$ to the next, and that of B does not change.

**The number of terms.**
The rows of $\Gamma$ sum to $\rho$, so $\mu$ is a right Perron vector, $\Gamma\mu = \rho\mu$,
and the $k$-th term is $\rho^k\mu$, of largest entry $0.5\rho^k$.
A stops at the first $k$ with $0.5\rho^k \leq 10^{-12}$,
$$k \geq \frac{\log(5\times10^{11})}{\log(1/\rho)} \approx \frac{27}{1-\rho},$$
with $\log\rho \approx -(1-\rho)$: the first such $k$ is 256, 2681 and 26925, the number of products A makes;
the approximation gives about 270, 2700 and 27000.
Each term is one product of a $2\times2$ matrix and a vector, at a fixed cost,
so the time of A grows about tenfold from one $\rho$ to the next.

**B** factorises and solves one $2\times2$ system, whatever $\rho$.

The verification notebook prints ratios of the time of A to that of B of about 100, 1000 and 10000,
and a growth of the time of A of about 10 at each rung, as the number of terms grows by 10.5 and then by 10.0.

```python
terms = log(5e11) / log(1 / 0.9), log(5e11) / log(1 / 0.99), log(5e11) / log(1 / 0.999)
log(5e11), terms, (ceil(terms[0]), ceil(terms[1]), ceil(terms[2])), terms[1] / terms[0], terms[2] / terms[1], log(1 / (1 - 0.9)), log(1 / (1 - 0.99)), log(1 / (1 - 0.999))
```

**The other choices.**
- (B) A faster, its products being cheaper than a factorisation: one product is, and thousands are not.
- (C) B faster by a factor that does not change: the number of terms taken not to depend on $\rho$;
  it grows as $1/(1-\rho)$.
- (D) The time of A growing by about the same amount from one $\rho$ to the next:
  the number of terms taken to grow as $\log(1/(1-\rho))$, which is $2.3$, $4.6$ and $6.9$;
  it grows as $1/\log(1/\rho) \approx 1/(1-\rho)$.
