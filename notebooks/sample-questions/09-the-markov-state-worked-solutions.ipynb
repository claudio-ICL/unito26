# Section 9 · The Markov state

**F1. The state, before and after an event.**
The decayed counts are $Z_e(t) = \sum_{j \,:\, T^e_j < t} e^{-\beta(t - T^e_j)}$, eq. (1.3.11, `eq.decayedCounts`).
The strict inequality makes $Z$ left-continuous:
at an arrival of type $e$, $Z(T)$ is the value before the jump
and $Z(T+) = Z(T) + \mathbf e_e$ the value after it
(the paragraph after eq. (1.3.13, `eq.intensityFromState`)).
The value before the jump enters the intensity $\lambda = \mu + AZ$;
the value after it is carried forward, Remark 1.3.41 (`remark.prePostJump`).
Between events $Z$ follows the flow $z \mapsto e^{-\beta u}z$, eq. (1.3.14, `eq.stateDynamics`), so that
$$Z(T_{n+1}) = \big(Z(T_n) + \mathbf e_{E_n}\big)\,e^{-\beta(T_{n+1} - T_n)},$$
eq. (1.3.38, `eq.update_of_exponential_intensity`).
`decayed_counts_at` carries out this update;
its test `times[event] < now` is the strict inequality,
so a clock time that falls on an event reads $Z$ before that event's jump.

**F2. The Markov property, homogeneous in time.**
By Theorem 1.3.14 (`thm.markovState`) $Z$ is a piecewise-deterministic Markov process
with the flow $e^{-\beta t}z$, the jump rate $\lambda_{\mathfrak g}(z) = \mathbf 1^\top(\mu + Az)$,
and the jump to $z + \mathbf e_e$ with probability $\lambda_e(z)/\lambda_{\mathfrak g}(z)$:
all three are functions of the state alone.
Hence, by the paragraph on homogeneity in time that follows the theorem,
conditionally on the history at $s$ the law of $\big(N(s+u) - N(s),\, Z(s+u)\big)_{u > 0}$
depends on $Z(s+)$ alone, and neither on $s$, nor on $N(s)$, nor on the path before $s$:
it is the law of the process started from $Z(s+)$.
An expectation over $(s, t]$ given $Z(s+) = z$ is a function of $z$ and $t - s$.
The same holds when $s$ is a stopping time,
a random time $\tau$ such that whether $\tau \le t$ is decided by the path up to $t$.

**F3. The forward means.**
With $K = A - \beta I$ and $\Phi(h) = \int_0^h e^{Ku}\,du$,
Proposition 1.3.33 (`prop.forwardMean`) gives
$$m_{s,s+h}(z) = \mathbb E\big[Z(s+h) \mid Z(s+) = z\big] = e^{Kh}z + \Phi(h)\,\mu,$$
eq. (1.3.30, `eq.forwardMeanState`), and
$$\mathbb E\big[N(s+h) - N(s) \mid Z(s+) = z\big] = h\mu + A\Phi(h)\,z + A\Big(\int_0^h \Phi(u)\,du\Big)\mu,$$
eq. (1.3.31, `eq.forwardMeanCounts`).
Both are affine in $z$ and depend on the window through $h$ alone.
If $\rho < 1$, with $m = \lambda^*/\beta$,
$m_{s,s+h}(z) - m = e^{Kh}(z - m)$, eq. (1.3.32, `eq.forwardMeanDeviation`),
and the expected counts are $h\lambda^* + A\Phi(h)(z - m)$, eq. (1.3.33, `eq.forwardMeanCountsDeviation`).
One more event of type $e'$ in the state displaces the mean state by $e^{Kh}\mathbf e_{e'}$,
Corollary 1.3.35 (`corol.meanResponse`), and so the expected counts by $A\Phi(h)\mathbf e_{e'}$.
If $\rho < 1$, as $h \to \infty$, $\mathbf 1^\top A\Phi(h)\mathbf e_{e'}$ tends to $C_{e'} - 1$,
the expected number of strict descendants, eq. (1.3.35, `eq.responseIntegral`),
where $C = \mathbf 1^\top(I - \Gamma)^{-1}$ are the cluster sizes, eq. (1.3.22, `eq.clusterSize`).

**F4. The one-way pair.**
$\mu = (0.4, 0.1)$, $A = \begin{pmatrix} 0.8 & 0 \\ 4 & 2.4 \end{pmatrix}$, $\beta = 4$.
The column sums of $A$ are $(4.8, 2.4)$.
$\Gamma = A/4 = \begin{pmatrix} 0.2 & 0 \\ 1 & 0.6 \end{pmatrix}$, eq. (1.3.12, `eq.GammaIsAOverBeta`),
is triangular, so its eigenvalues are its diagonal entries and $\rho = 0.6$.
Inverting the triangular $I - \Gamma = \begin{pmatrix} 0.8 & 0 \\ -1 & 0.4 \end{pmatrix}$,
$(I - \Gamma)^{-1} = \begin{pmatrix} 1/0.8 & 0 \\ 1/(0.8 \cdot 0.4) & 1/0.4 \end{pmatrix} = \begin{pmatrix} 1.25 & 0 \\ 3.125 & 2.5 \end{pmatrix}$,
$\lambda^* = (I - \Gamma)^{-1}\mu = (0.5, 1.5)$, eq. (1.3.18, `eq.stationaryIntensity`),
$m = \lambda^*/\beta = (0.125, 0.375)$, eq. (1.3.19, `eq.meanState`),
and $C = (4.375, 2.5)$.
$K = \begin{pmatrix} -3.2 & 0 \\ 4 & -1.6 \end{pmatrix}$ is triangular too, and
$$e^{Ku} = \begin{pmatrix} e^{-3.2u} & 0 \\ 2.5\,(e^{-1.6u} - e^{-3.2u}) & e^{-1.6u} \end{pmatrix},$$
the lower entry solving $x' = -1.6\,x + 4e^{-3.2u}$, $x(0) = 0$.
Every entry of $e^{Ku}$, hence of $\Phi(h)$ and of $A\Phi(h)$, is non-negative.
The slowest rate is $\beta(1 - \rho) = 1.6$:
the relaxation time is $0.625$ s, Definition 1.3.36 (`def.relaxationTime`).

```python
inverse = ((1 / 0.8, 0.0), (1 / 0.32, 1 / 0.4))
lam = (inverse[0][0] * 0.4, inverse[1][0] * 0.4 + inverse[1][1] * 0.1)
m = (lam[0] / 4, lam[1] / 4)
C = (inverse[0][0] + inverse[1][0], inverse[0][1] + inverse[1][1])
inverse, lam, m, C, 1 / (4 * (1 - 0.6))
```

## 9.1 · `question.restartTheStateAfterAMarketSell`

A buy at the time 1, sells at 2 and 3, a buy at 4, read at the times 3 and 5;
$\beta = \log 2$, so that the state halves in a unit of time.
Each restart starts at the sell of $s = 2$, on the events and the clock moved by $s$.

**Answer: (C).** `True`, `False`, `True`, `False`.

**The whole history.**
By eq. (1.3.11, `eq.decayedCounts`),
$$Z(3) = \big(\tfrac14,\ \tfrac12\big),
\qquad Z(5) = \big(\tfrac1{16} + \tfrac12,\ \tfrac18 + \tfrac14\big) = (0.5625,\ 0.375);$$
at the time 3 the sell of the time 3 is not counted, the inequality being strict (F1).

**The two states at $s$.**
`before` reads the clock time 2, which falls on the sell:
the test `times[event] < now` leaves the sell out,
and `before` is $Z(2) = (0.5, 0)$, the value before the jump (F1).
`after` is $Z(2+) = (0.5, 1)$.

**The restarts.**
`restart` starts from `z` at the time 0 and reads the shifted clock, 1 and 3.
It reproduces the whole history when it counts the sell of $s$ exactly once.
- Line 1, `before` with the events at or after $s$: the sell is kept, at the shifted time 0,
  which precedes the clock time 1 and is read. Counted once: `True`.
- Line 2, `before` with the events after $s$: the sell is lost,
  and the first reading is $(0.25, 0)$ against $(0.25, 0.5)$: `False`.
- Line 3, `after` with the events after $s$: counted once, in the state: `True`.
- Line 4, `after` with the events at or after $s$: counted twice,
  and the first reading is $(0.25, 1)$: `False`.

Lines 1 and 3 are the two readings of F1 at an event:
the state before the jump with the event still to come,
and the state after the jump with the event spent.

```python
half = exp(-log(2))
z_3 = (half**2, half)
z_5 = (half**4 + half, half**3 + half**2)
first_readings = ((0.5 * half, (0 + 1) * half), (0.5 * half, 0 * half),
                  (0.5 * half, 1 * half), (0.5 * half, (1 + 1) * half))
z_3, z_5, first_readings
```

**The other choices.**
- (A) `False`, `False`, `True`, `True`: takes an event at the shifted time 0 for one already in $Z(0+)$,
  and so not read: lines 1 and 2 then lack the sell, and line 4 is line 3.
  The loop reads every event before the clock time, the one at 0 included.
- (B) `False`, `True`, `False`, `False`: takes `before` for $Z(s+)$, reading `<` as `<=`.
  Then `before` holds the sell, line 2 counts it once, lines 1 and 3 twice and line 4 three times.
- (D) `False` on every line: takes the restart to keep the events and the clock unshifted;
  $z$ then stands at the time 0, not at $s$, and its part of every reading is decayed over two units of time too many:
  the buy it carries puts the buys off on every line.

## 9.2 · `question.millisecondClockKeepsTheState`

The one-way pair (F4) in seconds,
and the same specification on a clock in milliseconds, $c = 1000$:
$\mu$, $A$ and $\beta$ divided by $c$, the times multiplied by it.
A buy at 0.2 s and sells at 0.35 s and 0.9 s;
the state is read at 1 s, and the counts are forecast over the next half second.

**Answer: (A).** `False True`, then `False True`.

**The state.**
By Remark 1.3.11 (`remark.timeRescaling`) the process $t \mapsto N(t/c)$,
whose arrival times are $c\,T_j$,
is the exponential-kernel process with $\mu/c$, $A/c$ and $\beta/c$,
and its decayed counts are $Z(t/c)$.
Each term of eq. (1.3.11, `eq.decayedCounts`) keeps its exponent,
$$\frac{\beta}{c}\,(c\,t - c\,T_j) = \beta\,(t - T_j),$$
so `z_ms` is `z`,
$$z = \big(e^{-3.2},\ e^{-2.6} + e^{-0.4}\big) \approx (0.041,\ 0.745),$$
and not `z / 1000`. The first line is `False True`.

**The forecast.**
On the new clock $K$ becomes $K/c$, and at the horizon $ch$
$$\int_0^{ch} e^{Ku/c}\,du = c\,\Phi(h).$$
In eq. (1.3.31, `eq.forwardMeanCounts`) the three terms are then
$(ch)(\mu/c) = h\mu$, $(A/c)\,c\Phi(h)\,z = A\Phi(h)z$,
and, since $\int_0^{ch} c\,\Phi(u/c)\,du = c^2\int_0^h \Phi$,
$(A/c)\,c^2\big(\int_0^h\Phi\big)(\mu/c) = A\big(\int_0^h\Phi\big)\mu$.
The expected counts are numbers of events, the same over the same half second,
which on the new clock is the horizon 500;
the horizon 0.5 is half a millisecond.
The second line is `False True`.

```python
z = (exp(-4 * 0.8), exp(-4 * 0.65) + exp(-4 * 0.1))
z, (z[0] / 1000, z[1] / 1000)
```

**The other choices.**
- (B) `True False`, then `False False`: takes the state for a rate, divided by $c$ with $\mu$ and $A$;
  the forecast from $z/1000$ then misses at both horizons.
  The state is a sum of exponentials, each unchanged.
- (C) `False True`, then `True False`: reads the horizon `0.5` as half a second on either clock;
  on the clock in milliseconds a horizon is in milliseconds.
- (D) `False True`, then `False False`: takes the expected counts, like the intensities,
  to be divided by $c$ on the new clock; they are counts of events.

## 9.3 · `question.totalIntensityIsNotTheState`

Two states of equal total intensity on each of two specifications:
$(1, 0)$ and $(0.5, 0.5)$ on the self-exciting pair,
$\mu = (0.25, 0.25)$, $A = \begin{pmatrix} 2.8 & 0.2 \\ 0.2 & 2.8 \end{pmatrix}$, $\beta = 4$;
$(0.5, 0)$ and $(0, 1)$ on the one-way pair (F4).
Each line prints the gap of $\lambda_{\mathfrak g}$ and the gap of the expected total count over ten seconds.

**Answer: (B).** `0.00 0.00`, then `0.00 0.19`.

**The total intensity.**
$\lambda_{\mathfrak g} = \bar\mu + (\mathbf 1^\top A)\,z$ weights $z$ by the column sums of $A$,
the jumps of $\lambda_{\mathfrak g}$, Remark 1.3.10 (`remark.countsAgainstRates`):
$(3, 3)$ on the self-exciting pair and $(4.8, 2.4)$ on the one-way pair.
The gaps are $3 - (1.5 + 1.5) = 0$ and $0.5 \cdot 4.8 - 2.4 = 0$,
printed `0.00`; the option `z` prints a residue of rounding, of either sign, as `0.00`.

**The expected counts.**
In eq. (1.3.31, `eq.forwardMeanCounts`) only the term $A\Phi(h)z$ depends on $z$,
so the gap of the total counts is $\mathbf 1^\top A\Phi(10)(z_1 - z_2)$.
By F3 the weight $\mathbf 1^\top A\Phi(h)\mathbf e_{e'}$ tends to $C_{e'} - 1$;
the stem puts what is left at ten seconds at the order of $e^{-10}$ and $e^{-16}$,
so the weights are $C - 1$.
- On the self-exciting pair the column sums of $\Gamma$ are constant, $0.75 = \rho$,
  and Corollary 1.3.25 (`corol.constantClusterSize`) gives $C = (4, 4)$:
  $C - 1 = (3, 3)$, proportional to the column sums of $A$, and the gap is $3 \cdot 0.5 - 3 \cdot 0.5 = 0$.
- On the one-way pair $C - 1 = (3.375, 1.5)$ (F4), not proportional to $(4.8, 2.4)$:
  the gap is $0.5 \cdot 3.375 - 1.5 = 0.1875$, printed `0.19`.

$\lambda_{\mathfrak g}$ is the jump rate of Theorem 1.3.14 (`thm.markovState`), a function of the state, and not the state.
The two states of the one-way pair give the next event one rate
and different intensities, $\mu + Az = (0.8, 2.1)$ and $(0.4, 2.5)$,
and the first state lies in the buys, which leave more descendants per unit of jump of $\lambda_{\mathfrak g}$:
$3.375/4.8 \approx 0.70$, against $1.5/2.4 = 0.625$ for the sells.

```python
0.5 * 4.8 - 2.4, 0.5 * (C[0] - 1) - (C[1] - 1), (C[0] - 1) / 4.8, (C[1] - 1) / 2.4, 0.5 * C[0] - C[1]
```

**The other choices.**
- (A) `0.00 0.00` on both lines: takes $\lambda_{\mathfrak g}$ for the state,
  so that equal total intensities forecast equal counts.
- (C) `0.00 -0.19` on the second line: takes `more` for the count from `z2` less that from `z1`,
  the right size with the sign reversed.
- (D) `0.00 -0.31` on the second line: weights by the cluster sizes $C = (4.375, 2.5)$,
  which count the ancestor, Proposition 1.3.23 (`prop.clusterSize`):
  $0.5 \cdot 4.375 - 2.5 = -0.3125$.
  The ancestor is in the state, not among the events of the ten seconds.

## 9.4 · `question.fanStartsTheClockAtZero`

The one-way pair at $s = 30$, in the state $z = Z(s+) = (0.5, 2)$;
the expected counts over $(s, s+1]$, estimated by the mean of independent draws in four ways.

**Answer: (D).** `estimate_a` and `estimate_d`.

**What the simulator reads.**
In `ExponentialHawkes.step` the excess $D$, the two waits,
and in `_draw_type_and_jump` the type,
are functions of `decayed_counts` and of the uniforms.
`_advance` adds the wait to `time` and decays the state by it.
The time is accumulated and returned, and enters no draw:
the path after any time has a law fixed by the state at that time, as F2 states.

**The four estimates.**
- `estimate_a` starts from $z$ at the time 0 and counts the events of $(0, 1]$.
  By F2 this is the law of the counts over $(s, s+1]$ given $Z(s+) = z$:
  it tends to the expectation.
- `estimate_d` starts from $z$ with the clock set to 30.
  Its draws are those of `estimate_a`, every time moved by 30,
  and its count over $(30, 31]$ is the count of `estimate_a` over $(0, 1]$:
  it tends to the expectation.
- `estimate_b` starts from $z$ at the time 0 and counts over $(30, 31]$,
  thirty seconds after the path was at $z$.
- `estimate_c` starts empty and counts over $(30, 31]$.

From a state $z_0$ at the time 0,
eq. (1.3.33, `eq.forwardMeanCountsDeviation`) gives the expected count over $(30, 31]$ as
$\lambda^* + A\big(\Phi(31) - \Phi(30)\big)(z_0 - m)$,
and $\Phi(31) - \Phi(30) = e^{30K}\Phi(1)$ is of the order of $e^{-1.6 \cdot 30} = e^{-48}$ (F4).
Both `estimate_b` and `estimate_c` tend to $\lambda^* = (0.5, 1.5)$, up to a term of the order of $e^{-48}$:
the start is forgotten within a few relaxation times of 0.625 s.

**The expectation is not the stationary count.**
By eq. (1.3.33, `eq.forwardMeanCountsDeviation`) the expectation is $\lambda^* + A\Phi(1)(z - m)$,
with $z - m = (0.375, 1.625)$.
Every entry of $A\Phi(1)$ is non-negative, and its diagonal entries, $0.8\,\Phi_{0,0}(1)$ and $2.4\,\Phi_{1,1}(1)$, are positive (F4),
so both components exceed $(0.5, 1.5)$.
With $\Phi(1)$, the integral of $e^{Ku}$ of F4, the expectation is $(0.590, 4.343)$;
the run of `09-the-markov-state` prints means of $(0.49, 1.45)$ for `estimate_b`
and $(0.51, 1.56)$ for `estimate_c`.

```python
e32, e16 = exp(-3.2), exp(-1.6)
phi_1 = ((1 - e32) / 3.2, 0.0), (2.5 * ((1 - e16) / 1.6 - (1 - e32) / 3.2), (1 - e16) / 1.6)
d = (0.5 - m[0], 2.0 - m[1])
phi_d = (phi_1[0][0] * d[0], phi_1[1][0] * d[0] + phi_1[1][1] * d[1])
target = (lam[0] + 0.8 * phi_d[0], lam[1] + 4 * phi_d[0] + 2.4 * phi_d[1])
phi_1, target, exp(-1.6 * 30)
```

**The other choices.**
- (A) `estimate_a` only: takes the time of the simulator to enter its draws,
  so that setting it to 30 changes the law; no draw of `step` depends on it.
- (B) `estimate_a`, `estimate_b` and `estimate_d`: takes `estimate_b` to count from $z$;
  it counts thirty seconds later, when $z$ is forgotten.
- (C) `estimate_a`, `estimate_c` and `estimate_d`: takes a fresh path run to the time 30
  for one that is at $z$ there; nothing makes it so.
- (E) All four: takes the homogeneity in time of F2 for stationarity,
  as if every window of one second had one expected count whatever the state at its start;
  `estimate_b` and `estimate_c` forget $z$, and the expectation given $z$ does not.

## 9.5 · `question.forecastBetweenEventsFollowsTheFlow`

The one-way pair; a forecast made at $t = 10$, between events, of the counts over $(t, t+1]$,
from `last`, the time $T_n$ of the last event at or before $t$, and `state`, the state just after it.

**Answer: (A).** `forecast_a` only.

**What the loop records.**
At each yield of `events`, `step` has drawn the event and `_draw_type_and_jump` has added its unit vector:
`simulator.decayed_counts` is $Z(T_n+)$, and the copy keeps it.
A path with no event before $t$ keeps `last` $= 0$ and `state` $= 0 = Z(0+)$.
The event drawn past $t$ is held by `events` and yielded by the second call,
so `realised` counts every event of $(t, t+1]$.

**The state at $t$.**
No event falls in $(T_n, t]$, so by the flow of F1
$$Z(t+) = Z(t) = e^{-\beta(t - T_n)}\,Z(T_n+),$$
the state of `forecast_a`.
By F2 the forecast from $Z(t+)$ is the expectation of the counts given the history at $t$,
and the realised count less it has mean zero.

**The others.**
- `forecast_b` forecasts from $Z(T_n+)$ as if it were the state at $t$, omitting the flow.
  $Z(T_n+) \ge Z(t+)$ entrywise and the forecast increases with the state (F4),
  so it overstates: the run prints a mean error of $(-0.09, -2.24)$.
- `forecast_c` forecasts from $Z(T_n+)$ the window that opens $u = t - T_n$ later.
  The forecast being affine in the state (F3), this is the forecast from the mean state
  $m_{T_n,t}(Z(T_n+)) = e^{Ku}Z(T_n+) + \Phi(u)\mu$,
  an average over paths with further events in $(T_n, t]$, which this path has not had.
  The mean state exceeds $Z(t+) = e^{-\beta u}Z(T_n+)$ by $e^{-\beta u}(e^{Au} - I)Z(T_n+) + \Phi(u)\mu$,
  positive in both components,
  so `forecast_c` exceeds `forecast_a` on every path, in both components (F4).
  The run prints a mean error of $(-0.02, -0.96)$:
  the buys lie within the noise of the run there, and the argument decides them.
- `forecast_d` decays $Z(T_n+)$ from the open, by $e^{-\beta t} = e^{-40} \approx 4 \times 10^{-18}$,
  and forecasts from about the empty state: it understates, $(+0.04, +0.73)$.

```python
exp(-4 * 10)
```

**The other choices.**
- (B) `forecast_a` and `forecast_c`: takes the state after the last event for the information at $t$;
  that nothing has arrived over $(T_n, t]$ is information too, and $Z(t+)$ carries it.
- (C) `forecast_c` only: takes the forward means of F3 to hold only at an event, where $Z(s+)$ is read;
  they hold at every time.
- (D) `forecast_b` only: takes $Z$ to stay at $Z(T_n+)$ until the next event, as if it moved only at events;
  `forecast_b` is then the forecast from the state at $t$, and `forecast_a` decays it a second time.
  Between events $Z$ follows the flow of F1.

## 9.6 · `question.forecastAfterAMarketSellUsesThePostJumpState`

The one-way pair; a forecast of the counts over $(s, s+1]$ at the first market sell of each path, at $s$,
from the state that `decayed_counts_at` reads at $s$,
with or without the unit vector of the sell
($\mathbf e_1$ here, $\mathbf e_2$ in the key, which counts the components from 1).
$e^{-1.6} \approx 0.20$.

**Answer: (B).** Without the jump the forecast understates the sells by about 1.2 and not the buys;
with it, the forecast is unbiased.

**The state read at $s$.**
The clock time $s$ falls on the sell, and `decayed_counts_at` reads $Z(s)$, before the jump (F1).
The forecast with the jump starts from $Z(s+) = Z(s) + \mathbf e_1$.

**With the jump.**
The first sell is a stopping time: whether it has arrived by $t$ is decided by the path up to $t$.
By F2 the forward mean of F3 from $Z(s+)$ is then the expectation of the counts over $(s, s+1]$
given the history at $s$.
Every path has its first sell before 29, so the window lies within the path:
the forecast is unbiased.

**Without the jump.**
The forecast is affine in the state (F3),
and the two forecasts differ by $A\Phi(1)\mathbf e_1$,
the expected descendants in $(s, s+1]$ of the sell at $s$ (Corollary 1.3.35 (`corol.meanResponse`)).
The second column of $K$ is $(0, 2.4 - 4) = -1.6\,\mathbf e_1$,
so $e^{Ku}\mathbf e_1 = e^{-1.6u}\mathbf e_1$ and
$$A\Phi(1)\,\mathbf e_1 = \frac{1 - e^{-1.6}}{1.6}\,A\mathbf e_1
= \frac{1 - e^{-1.6}}{1.6}\begin{pmatrix} 0 \\ 2.4 \end{pmatrix} \approx \begin{pmatrix} 0 \\ 1.2 \end{pmatrix}.$$
The sells excite only the sells:
the buys are forecast right and the sells short by about 1.2.

```python
2.4 * (1 - exp(-1.6)) / 1.6, 2.4 * (1 - 0.20) / 1.6
```

**The other choices.**
- (A) Both unbiased: takes the sell at $s$, which the window $(s, s+1]$ leaves out,
  to leave the forecast unchanged; its descendants fall in the window.
- (C) Without the jump both understated: reads the kernel transposed, the rows for the causes.
  A sell would then excite the buys at 4, and the shortfall would be about $(2.4, 1.2)$.
- (D) With the jump the sells overstated: takes the state read at $s$ for $Z(s+)$,
  so that adding the unit vector counts the sell twice.

## 9.7 · `question.forecastFromAStoppingTime`

The check of 9.6, with the state after the jump,
at the event that each of four rules returns.

**Answer: (E).** `first_after_ten` and `first_after_a_lull`.

By F2 the forecast from $Z(\tau+)$ is unbiased when $\tau$ is a stopping time.
A rule that chooses the event by what comes after it is not one,
and the flow after the event it chooses differs from the forecast.
- `first_after_ten` returns the first event after the time 10, known when it arrives: a stopping time.

## 9.9 · `question.liveStateByRecursionAgainstRecomputation`

The state $Z(T_k+)$ after every event, by the update (A) and by the sum over the past (B),
timed on the first $n$ events of a path of the one-way pair, $n = 1000, \dots, 16000$.

**Answer: (D).** The time per event of A is about the same at every $n$,
and that of B grows several-fold from $n = 1000$ to $n = 16000$.

**What each computes.**
Both return $Z(T_k+)$: A adds the jump after the decay,
and the sum of B runs over $j \le k$, the event at $T_k$ entering with the weight $e^0 = 1$.

**The cost.**
A carries the state by eq. (1.3.38, `eq.update_of_exponential_intensity`) (F1),
a fixed amount of work per event.
B evaluates eq. (1.3.11, `eq.decayedCounts`) afresh at the $k$-th event,
over the $k$ events so far: about $n^2/2$ exponentials over $n$ events,
on average $n/2$ per event, 500 at $n = 1000$ and 8000 at $n = 16000$,
so that its time per event grows in proportion to $n$
once the fixed cost of each call is outweighed.
The sentence after eq. (1.3.38, `eq.update_of_exponential_intensity`) states the two costs,
$O(n^2)$ over the path against $O(1)$ per event.
numpy removes the loop of the interpreter over the $k$ terms, not the terms.
The run of `09-the-markov-state` prints a growth of the time per event,
from $n = 1000$ to $n = 16000$, of about 1 for A and of about 7 for B, varying from run to run,
against the factor 16 of the exponentials.

```python
1000 / 2, 16000 / 2, 16000 / 1000
```

**The other choices.**
- (A) B faster at every $n$: takes the call of numpy to remove the $k$ terms;
  it removes the loop of the interpreter over them.
- (B) Both about the same at every $n$, A faster by a constant factor:
  takes the work of B per event for a constant, one call of numpy each;
  the call computes $k$ exponentials.
- (C) Both grow several-fold: takes the update of A to read the history, as B does;
  it reads the last state and the last time.
